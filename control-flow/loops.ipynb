```python
#  For loop

'''range function generates a sequence of numbers, exclusive of the end value.'''

for i in range(5):
    print(i)
```

```python
# Step value in range function
# The step value in the range function allows you to specify the increment between each number in the sequence.

for i in range(1, 10, 2):
    print(i)
```

```python
# Print all numbers in descending order using range function

for i in range(10, 0, -1):
    print(i)
```

```python
# String traversal using for loop

string = "Hello, World!"
for char in string:
    print(char)
```

```python
# While loop example
# The while loop continues to execute as long as the condition is true.

count = 0
while count < 5:
    print(count)
    count += 1  # Increment the count to avoid an infinite loop
```

## Loop control statements
### Loop control statements allow you to alter the flow of a loop. The most common control statements `break`, `continue`, and `pass`.

```python
# Break statement
# The `break` statement is used to exit a loop prematurely when a certain condition is met.

for i in range(10):
    if i == 5:
        break  # Exit the loop when i is 5
    print(i)
```

```python
# Continue statement
# The `continue` statement skips the current iteration of a loop and moves to the next iteration.

for i in range(10):
    if i % 2 == 0:
        continue  # Skip even numbers
    print(i)  # This will only print odd numbers
```

```python

# Pass statement
# The `pass` statement is a null operation; it does nothing when executed. It is often used as a placeholder.

for i in range(5):
    if i == 2:
        pass  # Do nothing for i == 2
    print(i)  # This will print all numbers from 0 to 4, including 2
```

```python
# Nested loops
# Nested loops are loops inside another loop. The inner loop runs for each iteration of the outer

for i in range(3):  # Outer loop
    for j in range(2):  # Inner loop
        print(f"Outer loop {i}, Inner loop {j}")
```

```python
# Calculate the sum of numbers from 1 to 10 using a while loop

sum = 0
maxRange = 10
count = 1
while count <= maxRange:
    sum += count
    count += 1
print(f"The sum of numbers from 1 to 10 is: {sum}")
```

```python
# Calculate the sum of numbers from 1 to 10 using a for loop

sum = 0
maxRange = 10
for num in range(1, maxRange + 1):
    sum += num
print(f"The sum of numbers from 1 to 10 using for loop is: {sum}")
```

```python
#  Prime numbers bewtween 1 and 100 using a for loop

for num in range(1, 101):
    if num > 1:  # Prime numbers are greater than 1
        for i in range(2, num):
            if (num % i) == 0:
                break   # Not a prime number
        else:
            print(num) 
            
```
